```python
import os
import glob
import subprocess
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, GroupKFold, cross_validate, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor

# ==============================================================================
# 1. CLONE REPO (IF NOT PRESENT) & AUTO-LOAD DATASET
# ==============================================================================
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter.git"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.exists(REPO_DIR) and not os.path.exists("work"):
    subprocess.run(["git", "clone", REPO_URL], check=True)
if os.path.exists(REPO_DIR):
    os.chdir(REPO_DIR)

data_files = [
    f for f in (glob.glob("**/*.parquet", recursive=True) + glob.glob("**/*.csv", recursive=True))
    if "sample_data" not in f
]
if not data_files:
    raise FileNotFoundError("No .parquet or .csv dataset found. Upload your Week-5 file or check repo path.")

df = pd.read_parquet(data_files[0]) if data_files[0].endswith(".parquet") else pd.read_csv(data_files[0])

# ==============================================================================
# 2. CONFIGURE TARGET, GROUP & LEAKAGE-FREE FEATURES
# ==============================================================================
TARGET_COL = df.select_dtypes(include=[np.number]).columns[-1]
group_candidates = [c for c in df.columns if any(k in c.lower() for k in ["client", "domain", "site", "tenant"])]
GROUP_COL = group_candidates[0] if group_candidates else df.columns[0]

SAFE_FEATURES = [
    c for c in df.select_dtypes(include=[np.number]).columns
    if c not in {TARGET_COL, GROUP_COL} and not c.lower().endswith("_id")
]

df_clean = df.dropna(subset=[TARGET_COL, GROUP_COL]).reset_index(drop=True)
X, y, groups = df_clean[SAFE_FEATURES], df_clean[TARGET_COL], df_clean[GROUP_COL]

# ==============================================================================
# 3. BEFORE vs. AFTER VALIDATION AUDIT (NAIVE RANDOM vs. CLIENT-GROUPED)
# ==============================================================================
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1))
])

scoring = {"R2": "r2", "MAE": "neg_mean_absolute_error", "RMSE": "neg_root_mean_squared_error"}

naive_res = cross_validate(pipeline, X, y, cv=KFold(n_splits=5, shuffle=True, random_state=42), scoring=scoring)
honest_cv = GroupKFold(n_splits=min(5, groups.nunique()))
honest_res = cross_validate(pipeline, X, y, groups=groups, cv=honest_cv, scoring=scoring)

comparison = pd.DataFrame([
    {
        "Validation Strategy": "Before: Naive Random KFold",
        "Mean R2": round(naive_res["test_R2"].mean(), 4),
        "Mean MAE": round(-naive_res["test_MAE"].mean(), 4),
        "Mean RMSE": round(-naive_res["test_RMSE"].mean(), 4),
    },
    {
        "Validation Strategy": "After: Honest Client-Grouped KFold",
        "Mean R2": round(honest_res["test_R2"].mean(), 4),
        "Mean MAE": round(-honest_res["test_MAE"].mean(), 4),
        "Mean RMSE": round(-honest_res["test_RMSE"].mean(), 4),
    }
])

print(f"Dataset: {data_files[0]} | Target: {TARGET_COL} | Group: {GROUP_COL}\n")
print("=== BEFORE / AFTER VALIDATION COMPARISON ===")
print(comparison.to_string(index=False))

# ==============================================================================
# 4. TOP 5 WORST FAILURE EXAMPLES (OUT-OF-FOLD HONEST PREDICTIONS)
# ==============================================================================
df_clean["honest_pred"] = cross_val_predict(pipeline, X, y, groups=groups, cv=honest_cv)
df_clean["abs_error"] = (df_clean[TARGET_COL] - df_clean["honest_pred"]).abs()

print("\n=== TOP 5 WORST FAILURE EXAMPLES ===")
cols = [GROUP_COL, TARGET_COL, "honest_pred", "abs_error"] + SAFE_FEATURES[:3]
print(df_clean.sort_values("abs_error", ascending=False).head(5)[cols].to_string(index=False))
```
