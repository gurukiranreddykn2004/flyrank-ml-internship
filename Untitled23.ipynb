```python
import json
import os
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ==============================================================================
# 0. CONFIGURATION & PATHS
# ==============================================================================
INPUT_PATH = "work/outputs/w06_validated_predictions.csv"  # <-- Update to your W06 file
OUTPUT_QUEUE_PATH = "work/outputs/w07_action_playbook_queue.csv"
OUTPUT_FIG_PATH = "work/figures/w07_action_by_archetype.png"
OUTPUT_METRICS_PATH = "work/metrics/w07_playbook_summary.json"

os.makedirs("work/outputs", exist_ok=True)
os.makedirs("work/figures", exist_ok=True)
os.makedirs("work/metrics", exist_ok=True)

# Load your validated model output (or generate sample structure if testing)
if os.path.exists(INPUT_PATH):
    df = pd.read_csv(INPUT_PATH)
else:
    print(f"[INFO] {INPUT_PATH} not found. Using realistic placeholder schema for testing.")
    rng = np.random.default_rng(42)
    n = 150
    df = pd.DataFrame({
        "url": [f"/blog/post-{i}" for i in range(1, n + 1)],
        "archetype": rng.choice(["Evergreen Guide", "Product/Commercial", "News/Trending", "Thin/Legacy"], size=n, p=[0.4, 0.25, 0.2, 0.15]),
        "decay_score": np.round(rng.uniform(0.05, 0.98, size=n), 3),
        "monthly_traffic": rng.integers(10, 8000, size=n),
        "traffic_trend_60d": np.round(rng.uniform(-0.65, 0.30, size=n), 3),
        "is_ymyl_or_legal": rng.choice([False, True], size=n, p=[0.92, 0.08]),
    })

# ==============================================================================
# 1. ARCHETYPE -> ACTION MAPPING, REASON CODES & COST/VALUE RANKING
# ==============================================================================
def map_content_action(row):
    arch = row["archetype"]
    decay = row["decay_score"]
    traffic = row["monthly_traffic"]
    trend = row["traffic_trend_60d"]
    protected = row.get("is_ymyl_or_legal", False)

    # No-Go / Human-Escalation Gate first
    if protected and decay >= 0.65:
        return (
            "SME / Compliance Review Only",
            "RC_00_PROTECTED_YMYL",
            "High",
            "Mandatory SME sign-off; no automated edits",
        )

    # Archetype -> Action Logic
    if arch == "Evergreen Guide" and decay >= 0.70 and traffic >= 500:
        return (
            "Priority Content Refresh",
            "RC_01_HIGH_VALUE_EVERGREEN_DECAY",
            "High",
            "Editor SERP audit + factual/section refresh",
        )
    elif arch == "Product/Commercial" and decay >= 0.60:
        return (
            "Conversion & Intent Optimization",
            "RC_02_COMMERCIAL_DECAY",
            "High",
            "Product marketer + SEO review before update",
        )
    elif arch in ["Evergreen Guide", "Product/Commercial"] and 0.45 <= decay < 0.70:
        return (
            "Light Refresh & Internal Linking",
            "RC_03_MODERATE_DECAY_OPPORTUNITY",
            "Low",
            "Editor spot-check + internal link injection",
        )
    elif arch == "Thin/Legacy" and traffic < 100 and trend < -0.25:
        return (
            "Consolidate / Merge Candidate",
            "RC_04_THIN_DECLINING_CANDIDATE",
            "Medium",
            "Mandatory human backlink & redirect check",
        )
    elif arch == "News/Trending" and decay >= 0.75:
        return (
            "Archive or Historical Tagging",
            "RC_05_NATURAL_NEWS_DECAY",
            "Low",
            "Editor confirmation (do not force-refresh)",
        )
    else:
        return (
            "Monitor Only (No Action)",
            "RC_06_STABLE_PERFORMANCE",
            "None",
            "Automated tracking only",
        )

mapped = df.apply(map_content_action, axis=1, result_type="expand")
df[["recommended_action", "reason_code", "editorial_effort", "human_review_rule"]] = mapped

# Cost/Value Priority Score: favors high traffic recovery + high decay risk, penalizes heavy effort
effort_weight = {"None": 0.0, "Low": 1.2, "Medium": 0.85, "High": 1.0}
df["effort_multiplier"] = df["editorial_effort"].map(effort_weight)
df["value_priority_score"] = np.where(
    df["recommended_action"] == "Monitor Only (No Action)",
    0.0,
    np.round(np.log1p(df["monthly_traffic"]) * df["decay_score"] * df["effort_multiplier"], 3),
)

# Rank queue
df = df.sort_values(by=["value_priority_score", "decay_score"], ascending=False).reset_index(drop=True)
df["priority_rank"] = df.index + 1

# ==============================================================================
# 2. INTENDED USE, LIMITS, NO-GO LIST & MONITORING TRIGGERS (PLAYBOOK SPEC)
# ==============================================================================
playbook_governance = {
    "intended_use": (
        "Decision-support triage queue to help content editors prioritize weekly/monthly refresh sprints. "
        "Designed for non-production research and editorial planning—not autonomous publishing."
    ),
    "known_limits": [
        "Cannot observe upcoming product launches, brand voice shifts, or offline seasonality.",
        "Treats natural post-peak decay on News/Trending pieces differently from structural Evergreen loss, "
        "but relies on historical traffic accuracy.",
        "Does not measure backlink equity or downstream revenue attribution directly."
    ],
    "no_go_automation_list": [
        "NEVER auto-delete, unpublish, or 301-redirect URLs without human backlink/conversion verification.",
        "NEVER auto-publish LLM-generated rewrites directly to production.",
        "NEVER auto-modify YMYL, legal, medical, or compliance-regulated pages.",
        "NEVER trigger full rewrites on top-converting pages based solely on short-term (<14d) noise."
    ],
    "monitoring_and_retrain_triggers": {
        "editor_rejection_trigger": "Recalibrate thresholds if editors reject >30% of top-25 ranked actions.",
        "data_drift_trigger": "Retrain if input feature missingness exceeds 5% or traffic distribution shifts >25%.",
        "external_event_trigger": "Re-evaluate rules following any confirmed search engine core algorithm update.",
        "outcome_verification": "Audit refreshed URLs at 60 and 90 days post-update to measure traffic recovery."
    }
}

# ==============================================================================
# 3. EXPORTS FOR THE RESEARCH PAPER (QUEUE CSV, FIGURE & METRICS JSON)
# ==============================================================================
# Export 1: Ranked Queue CSV (stays in work/outputs/, ignored by git)
export_cols = [
    "priority_rank", "url", "archetype", "decay_score", "monthly_traffic",
    "traffic_trend_60d", "recommended_action", "reason_code",
    "editorial_effort", "value_priority_score", "human_review_rule"
]
df[export_cols].to_csv(OUTPUT_QUEUE_PATH, index=False)

# Export 2: Publication Figure (commit to work/figures/)
ct = pd.crosstab(df["archetype"], df["recommended_action"])
ax = ct.plot(kind="bar", stacked=True, figsize=(10, 5.5), colormap="viridis", edgecolor="black", linewidth=0.5)
ax.set_title("Content Action Playbook: Recommended Actions by Content Archetype", fontsize=12, fontweight="bold", pad=12)
ax.set_xlabel("Content Archetype", fontsize=10)
ax.set_ylabel("Number of URLs", fontsize=10)
plt.xticks(rotation=20, ha="right")
plt.legend(title="Recommended Action", bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8.5)
plt.tight_layout()
plt.savefig(OUTPUT_FIG_PATH, dpi=200, bbox_inches="tight")
plt.show()

# Export 3: Reproducible Summary JSON (commit to work/metrics/)
summary_receipt = {
    "generated_at_utc": datetime.utcnow().isoformat() + "Z",
    "total_urls_scored": int(len(df)),
    "actionable_urls": int((df["recommended_action"] != "Monitor Only (No Action)").sum()),
    "action_counts": df["recommended_action"].value_counts().to_dict(),
    "reason_code_counts": df["reason_code"].value_counts().to_dict(),
    "governance": playbook_governance,
}
with open(OUTPUT_METRICS_PATH, "w") as f:
    json.dump(summary_receipt, f, indent=2)

# ==============================================================================
# 4. SELF-CHECK ASSERTIONS
# ==============================================================================
assert os.path.exists(OUTPUT_QUEUE_PATH), "Missing exported queue CSV!"
assert os.path.exists(OUTPUT_FIG_PATH), "Missing exported figure PNG!"
assert os.path.exists(OUTPUT_METRICS_PATH), "Missing exported metrics JSON!"
assert df["reason_code"].isna().sum() == 0, "Null reason codes found!"
assert df["priority_rank"].is_monotonic_increasing, "Queue is not properly ranked!"

print("=== SELF-CHECK PASSED ===")
print(f"Queue exported  -> {OUTPUT_QUEUE_PATH} ({len(df)} rows)")
print(f"Figure exported -> {OUTPUT_FIG_PATH}")
print(f"Metrics receipt -> {OUTPUT_METRICS_PATH}\n")
print("Top 10 Ranked Actions Preview:")
print(df[export_cols[:9]].head(10).to_string(index=False))
```
