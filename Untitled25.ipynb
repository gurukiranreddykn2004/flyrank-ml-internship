```python
import re
from urllib.parse import urljoin, urlparse
import requests
from bs4 import BeautifulSoup

# ==============================================================================
# 1. CONFIGURATION
# ==============================================================================
OWNER_NAME = "Guru Kiran Reddy K N"
OUTPUT_REPORT = "LAUNCH_VERIFICATION.md"
HEADERS = {"User-Agent": "Mozilla/5.0 (compatible; LaunchHygieneVerifier/1.0)"}

site_url = input("Enter your live website URL (e.g. https://yourname.netlify.app): ").strip()
if not site_url.startswith(("http://", "https://")):
    site_url = "https://" + site_url

parsed = urlparse(site_url)
if not site_url or " " in site_url or "." not in parsed.netloc:
    print(f"\n[INVALID URL] '{site_url}' is not a valid website address.")
    print("Please paste your actual deployed website link (must contain a domain like .netlify.app, .vercel.app, .github.io, or .com).")
else:
    print(f"\nVerifying live deployment at: {site_url} ...\n")
    try:
        resp = requests.get(site_url, headers=HEADERS, allow_redirects=True, timeout=15)
        resp.raise_for_status()
    except Exception as e:
        print(f"[CONNECTION ERROR] Could not reach {site_url}: {e}")
    else:
        final_url = resp.url
        html = resp.text
        soup = BeautifulSoup(html, "html.parser")
        results = []

        # 2. HTTPS & LIVE DOMAIN CHECK
        is_https = final_url.startswith("https://")
        results.append(("Live Over HTTPS", is_https, f"Resolved to `{final_url}` (HTTP {resp.status_code})"))

        # 3. ANALYTICS SCRIPT CHECK
        analytics_signatures = {
            "Google Analytics (GA4 / gtag)": ["googletagmanager.com/gtag/js", "google-analytics.com"],
            "Cloudflare Web Analytics": ["static.cloudflareinsights.com"],
            "Plausible Analytics": ["plausible.io/js"],
            "Umami Analytics": ["umami.js", "analytics.umami.is"],
            "Vercel / Netlify / Simple Analytics": ["_vercel/insights", "simpleanalyticscdn.com", "netlify"],
        }
        detected_analytics = [
            name for name, patterns in analytics_signatures.items()
            if any(p in html.lower() for p in patterns)
        ]
        results.append((
            "Analytics Installed",
            len(detected_analytics) > 0,
            ", ".join(detected_analytics) if detected_analytics else "No known analytics script detected in HTML source"
        ))

        # 4. LAUNCH HYGIENE: PAGE TITLE, FAVICON & SOCIAL SHARE PREVIEW
        title_tag = soup.find("title")
        title_text = title_tag.get_text(strip=True) if title_tag else ""
        results.append((
            "Page Title",
            len(title_text) >= 8,
            f"`{title_text}`" if title_text else "Missing <title> tag"
        ))

        favicon_tag = soup.find("link", rel=lambda r: r and any(k in r.lower() for k in ["icon", "shortcut icon", "apple-touch-icon"]))
        favicon_ok = False
        favicon_detail = "Missing <link rel='icon'> tag"
        if favicon_tag and favicon_tag.get("href"):
            fav_url = urljoin(final_url, favicon_tag["href"].strip())
            try:
                fav_resp = requests.head(fav_url, headers=HEADERS, allow_redirects=True, timeout=8)
                favicon_ok = fav_resp.status_code < 400
                favicon_detail = f"`{fav_url}` (HTTP {fav_resp.status_code})"
            except Exception:
                favicon_detail = f"Unreachable favicon URL: `{fav_url}`"
        results.append(("Favicon Accessible", favicon_ok, favicon_detail))

        og_title = soup.find("meta", property="og:title")
        og_desc = soup.find("meta", property="og:description")
        og_image = soup.find("meta", property="og:image")
        og_ok = all(tag and tag.get("content", "").strip() for tag in [og_title, og_desc, og_image])
        og_detail = (
            f"og:title=`{og_title['content'].strip()}` | og:image=`{og_image['content'].strip()}`"
            if og_ok else "Missing one or more of `og:title`, `og:description`, `og:image`"
        )
        results.append(("Social-Share Preview (Open Graph)", og_ok, og_detail))

        # 5. FLYRANK GRADUATE BADGE IN FOOTER CHECK
        footer = soup.find("footer")
        badge_in_footer = False
        badge_detail = "No <footer> tag or FlyRank badge found in footer"
        if footer:
            footer_html = str(footer).lower()
            badge_link = footer.find("a", href=True)
            badge_img = footer.find("img", src=True)
            if ("internship-badge" in footer_html or "flyrank" in footer_html) and (badge_link or badge_img):
                badge_in_footer = True
                href_val = badge_link["href"] if badge_link else "[no link]"
                badge_detail = f"Badge found in `<footer>` linking to `{href_val}`"
        results.append(("FlyRank Graduate Badge in Footer", badge_in_footer, badge_detail))

        # 6. PRINT SUMMARY & WRITE SUBMISSION REPORT
        table_rows = ""
        for check_name, passed, detail in results:
            status = "PASS" if passed else "ACTION NEEDED"
            print(f"[{status}] {check_name}: {detail}")
            table_rows += f"| {check_name} | **{status}** | {detail} |\n"

        report_md = f"""# Plant Your Flag: Domain, Analytics & Badge Verification
**Owner:** {OWNER_NAME}
**Live URL:** {final_url}

## Launch Verification Checklist
| Requirement | Status | Verification Details |
|---|---|---|
{table_rows}
## Additional Evidence
- **Analytics Screenshot:** `[Attach screenshot of live analytics dashboard here]`
- **Mobile Device Check:** Opened `{final_url}` on mobile browser — layout, HTTPS lock icon, favicon, and footer badge verified.
"""
        with open(OUTPUT_REPORT, "w", encoding="utf-8") as f:
            f.write(report_md)

        print(f"\nReport saved -> {OUTPUT_REPORT}")
```
